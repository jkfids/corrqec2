### Storm Model Test

```python
from corrqec2.noisemodels import Storm
from corrqec2.experiments import SurfaceCodeMemory

experiment = SurfaceCodeMemory(distance=9, rounds='3d', memory_type='Z')
```

```python
a = 0.0001
b = 0.5
emissions = [[1., 0., 0., 0.], [0.25, 0.25, 0.25, 0.25]]

model_params = {
    'a': a,
    'b': b,
    'emissions': emissions,
}
p = 0.005
gate_noise = {
    #'after_identity_depolarization': p,
    'after_clifford_depolarization': p,
    'before_measure_flip_probability': p,
    #'after_reset_flip_probability': p,
}
noisy_qubit_types = 'all'
model = Storm(model_params=model_params, gate_noise=gate_noise, noisy_qubit_types=noisy_qubit_types)
```

```python
results = model.gen_error_matrix(experiment, n_samples=1000)
```

```python
import sinter
import stim

metadata = {
    'experiment_args': {'distance': 9, 'rounds': '3d', 'memory_type': 'Z'},
    'noise_model_args': {
        'model_params': model_params,
        'gate_noise': gate_noise,
        'noisy_qubit_types': noisy_qubit_types,
    },
    'min_batch_size': 1000,
}
task = sinter.Task(circuit=stim.Circuit(), json_metadata=metadata)
```

```python
sampler = SinterSampler(SurfaceCodeMemory, Storm)
```

```python
import jax

stats = sinter.collect(tasks=[task], num_workers=10, decoders='custom_sampler', custom_decoders={'custom_sampler': sampler}, max_shots=10000, print_progress=True)
```
